### Задача 1.73
> В данный район изделия поставляются тремя фирмами в соотношении 5:8:7. Среди продукции первой фирмы стандартные изделия составляют 90%, второй — 85%, третьей — 75%. Найти вероятность того, что: а) приобретённое изделие окажется нестандартным; б) приобретённое изделие оказалось стандартным. Какова вероятность того, что оно изготовлено третьей фирмой?

**Гипотезы:** $H_1$ — изделие от 1-й фирмы, $H_2$ — от 2-й, $H_3$ — от 3-й.

$$P(H_1) = \frac{5}{20}, \quad P(H_2) = \frac{8}{20}, \quad P(H_3) = \frac{7}{20}$$

Условные вероятности стандартного изделия:
$$P(S|H_1) = 0{,}90, \quad P(S|H_2) = 0{,}85, \quad P(S|H_3) = 0{,}75$$

#### а) Вероятность нестандартного изделия

$$P(\bar{S}) = 1 - P(S) = 1 - \sum_{i=1}^{3} P(H_i) \cdot P(S|H_i)$$

```python
# Априорные вероятности гипотез
pH1, pH2, pH3 = 5/20, 8/20, 7/20

# Условные вероятности стандартного изделия
pS_H1, pS_H2, pS_H3 = 0.90, 0.85, 0.75

# Полная вероятность стандартного изделия
pS = pH1 * pS_H1 + pH2 * pS_H2 + pH3 * pS_H3

# Вероятность нестандартного
round(1 - pS, 4)
```

#### б) Вероятность того, что стандартное изделие от третьей фирмы (формула Байеса)

$$P(H_3|S) = \frac{P(H_3) \cdot P(S|H_3)}{P(S)}$$

```python
# Формула Байеса: P(H3 | S)
round((pH3 * pS_H3) / pS, 4)
```

### Задача 1.74
> Два стрелка сделали по одному выстрелу в мишень. Вероятность попадания для первого стрелка равна 0,6, для второго — 0,3. В мишени оказалась одна пробоина. Найти вероятность того, что она принадлежит первому стрелку.

**Гипотезы:** $H_1$ — попал только 1-й, $H_2$ — попал только 2-й.

$$P(H_1) = p_1(1-p_2), \quad P(H_2) = (1-p_1)p_2$$

$$P(H_1|A) = \frac{p_1(1-p_2)}{p_1(1-p_2)+(1-p_1)p_2}$$

```python
p1, p2 = 0.6, 0.3

# Вероятности гипотез (одна пробоина)
h1 = p1 * (1 - p2)   # только 1-й попал
h2 = (1 - p1) * p2   # только 2-й попал

# Формула Байеса: пробоина принадлежит 1-му стрелку
round(h1 / (h1 + h2), 4)
```

### Задача 1.75
> Вся продукция цеха проверяется двумя контролёрами: первый проверяет 55% изделий, второй — остальные. Вероятность пропустить нестандартное изделие: 0,01 для первого, 0,02 для второго. Взятое наудачу изделие, маркированное как стандартное, оказалось нестандартным. Найти вероятность того, что оно проверялось вторым контролёром.

**Гипотезы:** $H_1$ — проверял 1-й, $H_2$ — проверял 2-й. Событие $A$ — нестандартное изделие пропущено.

$$P(H_1) = 0{,}55, \quad P(H_2) = 0{,}45, \quad P(A|H_1) = 0{,}01, \quad P(A|H_2) = 0{,}02$$

$$P(H_2|A) = \frac{P(H_2) \cdot P(A|H_2)}{P(H_1) \cdot P(A|H_1) + P(H_2) \cdot P(A|H_2)}$$

```python
pH1, pH2 = 0.55, 0.45
pA_H1, pA_H2 = 0.01, 0.02

# Полная вероятность события A
pA = pH1 * pA_H1 + pH2 * pA_H2

# Формула Байеса: P(H2 | A)
round((pH2 * pA_H2) / pA, 4)
```

### Задача 1.79
> В лифт семиэтажного дома на первом этаже вошли три человека. Каждый из них с одинаковой вероятностью выходит на любом из этажей, начиная со второго. Найти вероятность того, что все пассажиры выйдут: а) на четвёртом этаже; б) на одном и том же этаже; в) на разных этажах.

Каждый пассажир равновероятно выходит на одном из **6** этажей (2–7). Общее число исходов: $6^3 = 216$.

#### а) Все выходят на 4-м этаже

$$P(A) = \frac{1}{6^3} = \frac{1}{216}$$

```python
n_floors = 6  # этажи 2..7
n_people = 3
total = n_floors ** n_people  # 216

# а) все выходят на 4-м этаже — единственный исход
round(1 / total, 4)
```

#### б) Все выходят на одном и том же этаже

$$P(B) = \frac{6}{6^3} = \frac{1}{36}$$

```python
# б) все на одном этаже — 6 благоприятных исходов (по одному на каждый этаж)
round(n_floors / total, 4)
```

#### в) Все выходят на разных этажах

$$P(C) = \frac{A_6^3}{6^3} = \frac{6 \cdot 5 \cdot 4}{216}$$

```python
# в) все на разных этажах — размещения A(6,3) = 6·5·4 = 120
round(math.perm(n_floors, n_people) / total, 4)
```

## Отдельное задание: задача об урнах

> В одной урне 15 белых и 5 чёрных шаров, в другой — 6 белых и 1 чёрный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны, — белый. Решить классически и смоделировать.

### Классическое решение

**Гипотезы** $H_k$ — среди 3 перенесённых шаров ровно $k$ белых ($k = 0, 1, 2, 3$).

Урна 1: $N=20$ шаров, $W=15$ белых, $B=5$ чёрных. Вероятности гипотез (гипергеометрическое распределение):
$$P(H_k) = \frac{C_{15}^{k} \cdot C_{5}^{3-k}}{C_{20}^{3}}$$

После переноса во 2-й урне $7+3=10$ шаров, из них $6+k$ белых. Условная вероятность белого:
$$P(A|H_k) = \frac{6+k}{10}$$

**Формула полной вероятности:**
$$P(A) = \sum_{k=0}^{3} P(H_k) \cdot P(A|H_k)$$

```python
# Классическое решение
W1, B1, n = 15, 5, 3          # урна 1: белых, чёрных; переносим n шаров
N1 = W1 + B1                  # 20
W2_init, N2_after = 6, 10     # урна 2: белых до переноса; всего после переноса

p_A = 0.0
for k in range(n + 1):
    pH_k = sc.comb(W1, k) * sc.comb(B1, n - k) / sc.comb(N1, n)
    pA_Hk = (W2_init + k) / N2_after
    p_A += pH_k * pA_Hk

round(p_A, 4)
```

### Моделирование (Монте-Карло)

Проведём $N = 1\,000\,000$ экспериментов и оценим вероятность.

```python
rng = np.random.default_rng(42)
N_sim = 1_000_000

# Урна 1: 15 белых (1) и 5 чёрных (0)
urn1 = np.array([1] * 15 + [0] * 5)

white_count = 0
for _ in range(N_sim):
    drawn = rng.choice(urn1, size=3, replace=False)  # 3 шара из урны 1
    k = int(drawn.sum())                              # сколько белых
    urn2 = [1] * (6 + k) + [0] * (1 + (3 - k))      # урна 2 после переноса
    white_count += int(rng.choice(urn2))             # вынуть 1 шар

round(white_count / N_sim, 4)
```
