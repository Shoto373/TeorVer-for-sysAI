## Задание 2.9
В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того, что из 9 пакетов акций в результате торгов по первоначально заявленной цене: 1) не будут проданы 5 пакетов; 2) будет продано: а) менее 2 пакетов; б) не более 2 пакетов; в) хотя бы 2 пакета; г) наивероятнейшее число пакетов.

Дано: n = 9, p = 0.2, q = 0.8


1) Не будут проданы 5 пакетов -> значит продано ровно 4 пакета (из 9).
По формуле Бернулли:
P_9(4) = C(9, 4) * 0.2^4 * 0.8^5


```python
round(stats.binom.pmf(4, 9, 0.2), 4)
```

2а) Менее 2 пакетов (0 или 1):
P(m < 2) = P(0) + P(1)


```python
round(stats.binom.cdf(1, 9, 0.2), 4)
```

2б) Не более 2 пакетов (m <= 2):
P(m <= 2) = P(0) + P(1) + P(2)


```python
round(stats.binom.cdf(2, 9, 0.2), 4)
```

2в) Хотя бы 2 пакета (от 2 до 9):
P(m >= 2) = 1 - P(m < 2)


```python
round(1 - stats.binom.cdf(1, 9, 0.2), 4)
```

2г) Наивероятнейшее число пакетов m0:
np - q <= m0 <= np + p
9*0.2 - 0.8 <= m0 <= 9*0.2 + 0.2
1 <= m0 <= 2
Получается m0 = 1 и m0 = 2
P = P_9(1) + P_9(2)


```python
round(stats.binom.pmf(1, 9, 0.2) + stats.binom.pmf(2, 9, 0.2), 4)
```

## Задание 2.11
По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины: а) 480 предприятий; б) наивероятнейшее число предприятий; в) не менее 480; г) от 480 до 520.

n = 1000, p = 0.5, q = 0.5
np = 500, npq = 250, sqrt(npq) = sqrt(250)
Используем теоремы Муавра-Лапласа


а) Ровно 480 предприятий (локальная теорема Муавра-Лапласа):
x = (480 - 500) / sqrt(250) ~ -1.265
P_1000(480) ~ phi(x) / sqrt(npq)


```python
n = 1000
p = 0.5
s = math.sqrt(n * p * (1 - p))
x = (480 - n * p) / s
round(stats.norm.pdf(x) / s, 4)
```

б) Наивероятнейшее число предприятий:
np - q <= m0 <= np + p => 500 - 0.5 <= m0 <= 500 + 0.5 => m0 = 500
x = (500 - 500) / sqrt(250) = 0


```python
x = (500 - n * p) / s
round(stats.norm.pdf(x) / s, 4)
```

в) Не менее 480 предприятий (от 480 до 1000) (интегральная теорема):
P(480 <= m <= 1000) = Ф(x2) - Ф(x1)


```python
x1 = (480 - n * p) / s
x2 = (1000 - n * p) / s
round(stats.norm.cdf(x2) - stats.norm.cdf(x1), 4)
```

г) От 480 до 520 предприятий (интегральная теорема):
P(480 <= m <= 520) = Ф(x2) - Ф(x1)


```python
x1 = (480 - n * p) / s
x2 = (520 - n * p) / s
round(stats.norm.cdf(x2) - stats.norm.cdf(x1), 4)
```

## Задание 2.22
В вузе 3650 студентов. Вероятность дня рождения в заданный день года равна 1/365. Найти а) наивероятнейшее число студентов, родившихся 1 мая, и вероятность этого числа, б) вероятность того, что хотя бы у трёх студентов совпадёт день рождения.

Дано n = 3650, p = 1/365, q = 364/365. Считаем дни рождения студентов независимыми. Так как n большое, а p малое, используем формулу Пуассона с lambda = n * p = 10.


а) Наивероятнейшее число студентов, родившихся 1 мая.

np - q <= m0 <= np + p => 3286/365 <= m0 <= 3651/365 => m0 = 10.

По формуле Пуассона P_3650(10) ~ 10^10 * e^(-10) / 10!


```python
print(f'По формуле Пуассона: {stats.poisson.pmf(10, 10):.4f}')
print(f'По формуле Бернулли: {stats.binom.pmf(10, 3650, 1 / 365):.4f}')
```

б) Если считать студентов, родившихся в заданный день (например, 1 мая), то P(m >= 3) = 1 - (P(0) + P(1) + P(2)).

Если дата может быть любой, совпадение гарантировано. При двух студентах на каждую дату было бы не более 2 * 365 = 730 студентов, а их 3650. В этом случае P = 1.


```python
print(f'По формуле Пуассона: {1 - stats.poisson.cdf(2, 10):.4f}')
print(f'По формуле Бернулли: {1 - stats.binom.cdf(2, 3650, 1 / 365):.4f}')
```

## Задание 2.27
С первого раза аудиторную работу по теории вероятностей успешно выполняют 50% студентов. Найти вероятность того, что из 400 студентов её успешно выполнят а) ровно 180 студентов, б) не менее 180 студентов.

Дано n = 400, p = 0.5, q = 0.5. Получаем np = 200, npq = 100, sqrt(npq) = 10. Используем формулы Муавра–Лапласа.


а) Ровно 180 студентов (локальная формула Муавра–Лапласа).

x = (180 - 200) / 10 = -2

P_400(180) ~ phi(-2) / 10


```python
n = 400
p = 0.5
s = math.sqrt(n * p * (1 - p))
x = (180 - n * p) / s
print(f'По формуле Муавра–Лапласа: {stats.norm.pdf(x) / s:.4f}')
print(f'По формуле Бернулли: {stats.binom.pmf(180, n, p):.4f}')
```

б) Не менее 180 студентов (интегральная формула Муавра–Лапласа).

x1 = (180 - 200) / 10 = -2, x2 = (400 - 200) / 10 = 20

P(180 <= m <= 400) ~ Ф(x2) - Ф(x1), где Ф — функция распределения стандартной нормальной величины.

По схеме Бернулли P(m >= 180) = 1 - P(m <= 179).


```python
x1 = (180 - n * p) / s
x2 = (400 - n * p) / s
print(f'По формуле Муавра–Лапласа: {stats.norm.cdf(x2) - stats.norm.cdf(x1):.4f}')
print(f'По формуле Бернулли: {1 - stats.binom.cdf(179, n, p):.4f}')
```

## Моделирование по схеме Бернулли
Для каждого набора n и p моделируем 1 000 000 серий. Число успехов m в n независимых испытаниях генерируем по биномиальному распределению. Частота события равна доле серий, в которых выполнено его условие.

В 2.22(б) моделируем число дней рождения в заданную дату. Если дата может быть любой, в каждой серии хотя бы трое студентов имеют общий день рождения.


```python
import numpy as np

rng = np.random.default_rng(42)
N = 1_000_000

m = rng.binomial(3650, 1 / 365, size=N)
print(f'2.22а) m = 10: {(m == 10).mean():.4f}')
print(f'2.22б) m >= 3 в заданную дату: {(m >= 3).mean():.4f}')

m = rng.binomial(400, 0.5, size=N)
print(f'2.27а) m = 180: {(m == 180).mean():.4f}')
print(f'2.27б) m >= 180: {(m >= 180).mean():.4f}')

m = rng.binomial(10000, 0.0001, size=N)
print(f'2.23а) m = 5: {(m == 5).mean():.4f}')
print(f'2.23б) m <= 2: {(m <= 2).mean():.4f}')

m = rng.binomial(10, 0.6, size=N)
print(f'2.25а) m = 6: {(m == 6).mean():.4f}')
m = rng.binomial(200, 0.6, size=N)
print(f'2.25б) m = 120: {(m == 120).mean():.4f}')

m = rng.binomial(800, 0.01, size=N)
print(f'2.30) m = 8: {(m == 8).mean():.4f}')
```
