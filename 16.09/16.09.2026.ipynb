### Задача 1.38
> Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

$P(A) = \frac{1}{5!}$

```python
round(1 / sc.factorial(5), 4)
```

* Решение кодом (перебор всех перестановок через `itertools`):

```python
from itertools import permutations

books = list(permutations('12345'))

"""
    Книги стоят слева направо в порядке нумерации томов (от 1 до 5)
"""
choice_books = [p for p in books if p == ('1', '2', '3', '4', '5')]
print(f'Всего перестановок: {len(books)}')
print(f'Книги стоят по порядку в {len(choice_books)} случаях.')
print(f'Вероятность: {len(choice_books) / len(books):.4f}')
```

### Задача 1.40
> Из 20 филиалов Сбербанка 10 расположены за чертой города. Для обследования случайным образом отобрано 5 филиалов. Какова вероятность того, что среди отобранных окажется в черте города: а) 3 филиала; б) хотя бы один?

а) $P(A) = \frac{C_{10}^3 \cdot C_{10}^2}{C_{20}^5}$

```python
round(sc.comb(10, 3) * sc.comb(10, 2) / sc.comb(20, 5), 4)
```

б) $P(B) = 1 - \frac{C_{10}^5}{C_{20}^5}$

```python
round(1 - sc.comb(10, 5) / sc.comb(20, 5), 4)
```

### Задача 1.43
> Наудачу взятый телефонный номер состоит из 5 цифр. Какова вероятность того, что в нем все цифры: а) различные; б) одинаковые; в) нечетные? Известно, что номер телефона не начинается с цифры ноль.

а) $P(A) = \frac{9 \cdot 9 \cdot 8 \cdot 7 \cdot 6}{9 \cdot 10^4}$

```python
round((9 * 9 * 8 * 7 * 6) / 90000, 4)
```

б) $P(B) = \frac{9}{9 \cdot 10^4} = \frac{1}{10000}$

```python
round(9 / 90000, 4)
```

в) $P(C) = \frac{5^5}{9 \cdot 10^4}$

```python
round(5**5 / 90000, 4)
```

* Решение кодом (перебор всех пятизначных номеров через `itertools`):

```python
from itertools import product

# Номера из 5 цифр, не начинающиеся с 0
first_digits = '123456789'
other_digits = '0123456789'

all_numbers = [
    d1 + ''.join(rest)
    for d1 in first_digits
    for rest in product(other_digits, repeat=4)
]
print(f'Всего возможных номеров: {len(all_numbers)}')

"""
    а) В номере все цифры различные
"""
choice_diff = [num for num in all_numbers if len(set(num)) == 5]
print(f'а) Все цифры различны в {len(choice_diff)} случаях.')
print(f'   Вероятность: {len(choice_diff) / len(all_numbers):.4f}')

"""
    б) В номере все цифры одинаковые
"""
choice_same = [num for num in all_numbers if len(set(num)) == 1]
print(f'б) Все цифры одинаковы в {len(choice_same)} случаях.')
print(f'   Вероятность: {len(choice_same) / len(all_numbers):.4f}')

"""
    в) В номере все цифры нечетные
"""
odd_digits = set('13579')
choice_odd = [num for num in all_numbers if all(d in odd_digits for d in num)]
print(f'в) Все цифры нечетны в {len(choice_odd)} случаях.')
print(f'   Вероятность: {len(choice_odd) / len(all_numbers):.4f}')
```

### Задача 1.48
> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.

а) $P(A) = \frac{27}{6^3} = \frac{1}{8}$

```python
round(27 / 6**3, 4)
```

б) $P(B) = \frac{108}{6^3} = \frac{1}{2}$

```python
round(108 / 6**3, 4)
```

* Решение кодом (перебор всех исходов через `itertools`):

```python
from itertools import product

ls = list(product('123456', repeat=3))

"""
    При бросании трех костей выпало 11 очков
"""
choice_11 = [el for el in ls if sum(list(map(int, el)))==11]
print(f'Выпало 11 очков в {len(choice_11)} случаях.')
print(*choice_11, sep='\n')

"""
    При бросании трех костей выпало больше 10 очков
"""
cnt = 0
for sm in range(11, 18+1):
    choice_sm = [el for el in ls if sum(list(map(int, el))) == sm]
    cnt += len(choice_sm)
print(f'Выпало более 10 очков в {cnt} случаях.')
```
